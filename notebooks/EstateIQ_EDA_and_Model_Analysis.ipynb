```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Load dataset
df = pd.read_csv("/content/india_housing_prices.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
# Display first five records
display(df.head())

# Display last five records
display(df.tail())

# Display column names
print("Columns:")
print(df.columns.tolist())

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# Missing values
print("Missing Values:")
display(df.isnull().sum())

# Duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# Data types
print("\nData Types:")
display(df.dtypes)

# Unique values in important categorical columns
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}: {df[column].nunique()} unique values")
    print(df[column].value_counts().head(10))
```

```python
# 1. Inspect rows containing missing values
print("Rows containing missing values:")
display(df[df.isnull().any(axis=1)])

# 2. Check floor-number inconsistencies
invalid_floors = df[df["Floor_No"] > df["Total_Floors"]]

print("\nRecords where Floor_No > Total_Floors:")
print("Count:", len(invalid_floors))
display(invalid_floors.head(10))

# 3. Check age and construction year relationship
age_difference = (
    df["Age_of_Property"] - (2025 - df["Year_Built"])
).abs()

print("\nAge consistency check:")
print("Records with inconsistent age:",
      (age_difference > 0).sum())

# 4. Check price-per-square-foot calculation
expected_price_per_sqft = (
    df["Price_in_Lakhs"] / df["Size_in_SqFt"]
)

price_difference = (
    df["Price_per_SqFt"] - expected_price_per_sqft
).abs()

print("\nPrice per square foot consistency:")
print("Average difference:", price_difference.mean())
print("Maximum difference:", price_difference.max())

# 5. Check basic numerical ranges
numerical_columns = [
    "BHK",
    "Size_in_SqFt",
    "Price_in_Lakhs",
    "Price_per_SqFt",
    "Year_Built",
    "Floor_No",
    "Total_Floors",
    "Age_of_Property",
    "Nearby_Schools",
    "Nearby_Hospitals"
]

display(df[numerical_columns].describe().round(2))
```

```python
# Distribution of floor inconsistencies
print("Floor_No > Total_Floors:")
print((df["Floor_No"] > df["Total_Floors"]).sum())

print("\nFloor_No == Total_Floors:")
print((df["Floor_No"] == df["Total_Floors"]).sum())

print("\nFloor_No < Total_Floors:")
print((df["Floor_No"] < df["Total_Floors"]).sum())

# Check unusual floor combinations
print("\nFloor combinations:")
display(
    df.groupby(["Floor_No", "Total_Floors"])
    .size()
    .reset_index(name="Count")
    .sort_values("Count", ascending=False)
    .head(15)
)

# Inspect records where floor number exceeds total floors
print("\nSample inconsistent records:")
display(
    df.loc[
        df["Floor_No"] > df["Total_Floors"],
        ["Property_Type", "City", "Floor_No", "Total_Floors"]
    ].head(20)
)
```

```python
# Inspect the last record
display(df[df["Price_in_Lakhs"].isnull()])
```

```python
# Create a separate cleaned copy
cleaned_df = df.dropna().copy()

# Check the results
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", cleaned_df.shape)

print("\nMissing values after cleaning:")
print(cleaned_df.isnull().sum().sum())

print("\nDuplicate rows after cleaning:")
print(cleaned_df.duplicated().sum())

# Remove unreliable columns from model dataset
model_df = cleaned_df.drop(
    columns=["ID", "Floor_No", "Total_Floors", "Price_per_SqFt"]
)

print("\nDataset prepared for further analysis:")
print(model_df.shape)

display(model_df.head())
```

```python
# Recalculate property age using 2026 as the reference year
cleaned_df["Age_of_Property"] = 2026 - cleaned_df["Year_Built"]

# Update the modelling dataset
model_df = cleaned_df.drop(
    columns=["ID", "Floor_No", "Total_Floors", "Price_per_SqFt"]
).copy()

print("Age updated successfully!")

display(
    cleaned_df[
        ["Year_Built", "Age_of_Property"]
    ].head(10)
)

print("\nUpdated age statistics:")
display(cleaned_df["Age_of_Property"].describe())
```

```python
import pandas as pd
import numpy as np

df_ml = pd.read_excel(
    "/content/merged_houseprice_regression_classification.xlsx",
    sheet_name="Regression+Class"
)

print("Dataset Shape:", df_ml.shape)

print("\nColumns:")
print(df_ml.columns.tolist())

print("\nMissing Values:")
print(df_ml.isnull().sum())

print("\nDataset Information:")
df_ml.info()

print("\nFirst 5 Rows:")
display(df_ml.head())
```

```python
# 1. Check duplicate records and property IDs
print("Duplicate rows:", df_ml.duplicated().sum())
print("Duplicate Property IDs:", df_ml["Property_ID"].duplicated().sum())

# 2. Check investment category distribution
print("\nInvestment Category Distribution:")
display(df_ml["Price_Category"].value_counts())

print("\nCategory percentages:")
display(
    df_ml["Price_Category"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

# 3. Check future price growth
price_columns = [
    "Current_Price",
    "Price_Year_1",
    "Price_Year_2",
    "Price_Year_3",
    "Price_Year_4",
    "Price_After_5_Years"
]

print("\nAverage annual growth rates:")

for i in range(1, len(price_columns)):
    previous = df_ml[price_columns[i - 1]]
    current = df_ml[price_columns[i]]

    growth = ((current / previous) - 1) * 100

    print(f"{price_columns[i]}: {growth.mean():.2f}%")

# 4. Check relationship between final price and category
print("\nAverage five-year price by category:")
display(
    df_ml.groupby("Price_Category")["Price_After_5_Years"]
    .agg(["count", "mean", "min", "max"])
)

# 5. Check correlations with future price
print("\nCorrelation with five-year price:")
display(
    df_ml.select_dtypes(include=np.number)
    .corr()["Price_After_5_Years"]
    .sort_values(ascending=False)
)
```

```python
# Check whether future price follows a fixed formula
df_ml["Growth_Ratio"] = (
    df_ml["Price_After_5_Years"] / df_ml["Current_Price"]
)

print("Five-year growth ratio:")
display(df_ml["Growth_Ratio"].describe())

# Check whether the target follows 6.4% annual compounding
expected_future_price = (
    df_ml["Current_Price"] * (1.064 ** 5)
)

difference = (
    df_ml["Price_After_5_Years"] - expected_future_price
).abs()

print("\nDifference from assumed 6.4% growth formula:")
print("Mean difference:", difference.mean())
print("Maximum difference:", difference.max())

# Inspect category boundaries
print("\nFive-year price ranges by category:")
display(
    df_ml.groupby("Price_Category")["Price_After_5_Years"]
    .agg(["min", "max", "mean", "count"])
)
```

```python
# Define input features
feature_columns = [
    "Area_sqft",
    "Bedrooms",
    "Bathrooms",
    "Floors",
    "Year_Built",
    "Location_Score",
    "School_Score",
    "Current_Price"
]

X = df_ml[feature_columns].copy()

# Classification target
y_class = df_ml["Price_Category"].copy()

# Regression target
y_reg = df_ml["Price_After_5_Years"].copy()

print("Input features shape:", X.shape)
print("Classification target shape:", y_class.shape)
print("Regression target shape:", y_reg.shape)

print("\nInput features:")
display(X.head())

print("\nClassification target distribution:")
display(y_class.value_counts())

print("\nRegression target statistics:")
display(y_reg.describe())
```

```python
# Compare five-year future price across categories

print("Price range by category:")
print(
    df_ml.groupby("Price_Category")["Price_After_5_Years"]
    .agg(["min", "max", "mean", "count"])
)

print("\nPrice quantiles:")
print(
    df_ml["Price_After_5_Years"].quantile([0.33, 0.66])
)
```
